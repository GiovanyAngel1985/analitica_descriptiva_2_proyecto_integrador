# 02. Consolidación de módulos 2025

Este notebook une los 12 meses por módulo y genera archivos anuales procesados.

```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
sys.path.append(str(PROJECT_ROOT / 'src'))

from config import DATA_RAW, ANNUAL_MODULES, EXPECTED_MODULES
from utils import listar_carpetas_mensuales, buscar_archivo_por_patrones, leer_csv_seguro, estandarizar_columnas, exportar_csv

pd.set_option('display.max_columns', 200)
```

```python
meses = listar_carpetas_mensuales(DATA_RAW)
modulos_objetivo = ['caracteristicas', 'fuerza_trabajo', 'ocupados']
modulos_objetivo
```

```python
def consolidar_modulo(nombre_modulo):
    frames = []
    for mes in meses:
        archivo = buscar_archivo_por_patrones(mes, EXPECTED_MODULES[nombre_modulo])
        if archivo is None:
            print(f'No encontrado en {mes.name}: {nombre_modulo}')
            continue
        df = leer_csv_seguro(archivo)
        df = estandarizar_columnas(df)
        df['MES_CARPETA'] = mes.name
        df['ARCHIVO_ORIGEN'] = archivo.name
        frames.append(df)
    if not frames:
        return None
    return pd.concat(frames, ignore_index=True)
```

```python
resultados = {}
for modulo in modulos_objetivo:
    df_mod = consolidar_modulo(modulo)
    resultados[modulo] = df_mod
    if df_mod is not None:
        salida = ANNUAL_MODULES / f'geih_2025_{modulo}.csv'
        exportar_csv(df_mod, salida)
        print(modulo, df_mod.shape, '->', salida)
```
