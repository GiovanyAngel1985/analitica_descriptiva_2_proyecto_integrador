# 01. Revisión de archivos

Este notebook valida la estructura de carpetas mensuales, inventaria archivos y prueba la lectura inicial de los CSV.

```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
sys.path.append(str(PROJECT_ROOT / 'src'))

from config import DATA_RAW, EXPECTED_MONTHS, EXPECTED_MODULES
from utils import listar_carpetas_mensuales, inventario_archivos, buscar_archivo_por_patrones, leer_csv_seguro

pd.set_option('display.max_columns', 200)
pd.set_option('display.max_rows', 200)
print(PROJECT_ROOT)
```

```python
meses = listar_carpetas_mensuales(DATA_RAW)
pd.DataFrame({'mes_detectado': [m.name for m in meses]})
```

```python
faltantes = sorted(set(EXPECTED_MONTHS) - set([m.name for m in meses]))
print('Meses faltantes:', faltantes if faltantes else 'Ninguno')
```

```python
inv = inventario_archivos(DATA_RAW)
inv.head(20)
```

```python
control = []
for mes in meses:
    fila = {'mes': mes.name}
    for modulo, patrones in EXPECTED_MODULES.items():
        archivo = buscar_archivo_por_patrones(mes, patrones)
        fila[modulo] = archivo.name if archivo else None
    control.append(fila)
control_df = pd.DataFrame(control)
control_df
```
